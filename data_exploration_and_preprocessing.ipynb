# Regras de Negócio

Para trabalhar com este dataset, é importante destacar as Regras de Negócio que esta escola adota.

- Para o 7° ano, as disciplinas ensinadas são: Matemática, Português, Ciências, História, Geografia e Inglês.
- A carga horária é dividida da seguinte forma:
  1. Matemática e Português: 200 aulas para cada matéria.
  2. Ciências: 120 aulas.
  3. História, Geografia e Inglês: 80 aulas para cada matéria.
- O ano letivo é dividido em 4 bimestres.
- A nota de um estudante em cada bimestre varia de 0 a 10.
- A aprovação de um estudante em uma disciplina depende de dois fatores:
  1.   Nota maior ou igual a 6.0.
  2.   Frequência maior ou igual a 75%.
- Caso a nota ou a frequência não atinjam os resultados esperados, o estudante é reprovado na disciplina.

# Pré-processamento

O pré-processamento é uma das etapas mais importantes para ser realizada quando se trata de analisar conjuntos de dados.

Nesta etapa, aplica-se as seguintes tarefas:

- Limpeza de dados: remoção de dados duplicados, correção de erros de digitação, padronização e preenchimento ou remoção de informações nulas.
- Transformação: transformação e conversão de valores e tipos.
- Redução da base de dados: obter as informações que serão cruciais para a análise, evitando levar informações que não são importantes para o objetivo final do trabalho.

A partir desta etapa, deixaremos o conjunto de dados quase pronto para realizar a escrita do arquivo e posteriormente a exportação para o Power BI, onde serão feitos os relatórios.

## Definição dos processos

A partir da análise exploratória dos dados e identificação dos problemas dentro do DataFrame, foram definidas as tarefas de pré-processamento a serem executadas dentro deste projeto:

1. **Padronização**: ajustar as escritas para manter o padrão, evitando diversas formas de palavras para itens que representam possuem o mesmo significado.
2. **Ajustar os tipos de dados**: modificar os tipos de dados para ficar de acordo com a informação a ser passada. Ex: tipo object deve ser alterado para int64 ou float64 caso a coluna informe valores numéricos.
3. **Validação de valores**: verificar se os valores numéricos estão de acordo com as regras de validação da base de dados.
4. **Tratamento de valores nulos**: analisar as informações nulas e sem registro e realizar as devidas alterações, mantendo a integridade da base de dados.
5. **Remoção de duplicatas**: valores duplicados não podem estar na tabela, portanto é necessário excluí-los para assim diminuir a base de dados e não poluir com valores repetidos.
6. **Verificação da qualidade do DataFrame**: realizar novamente e de forma mais resumida a Análise Exploratória do conjunto de dados para verificar se o DataFrame está dentro dos padrões esperados.

### Disciplinas

Neste caso, o tratamento desta coluna foi diferente. Não existe em uma tabela ou outra coluna que possui algo relacionado as Disciplinas, apenas no DataFrame notas. Foi preciso identificar quais disciplinas estão presentes nesta coluna, identificar as variações de escritas e padronizá-las.

```python
df_notas['Disciplina'].unique()
```

Disciplinas identificadas: Português, Matemática, Ciências, História, Geografia e Inglês.

Porém, há 26 grafias diferentes para se referir as disciplinas. Foi preciso criar um dicionário para cada disciplina e armazenar todas as variações de escrita.

```python
grafias_disciplinas = {
    'Português': ['Português', 'portugues', 'PORT.', 'PORTUGUES', 'Língua Portuguesa'],
    'Matemática': ['Matemática', 'MATEMATICA', 'matemática', 'MAT.'],
    'Ciências': ['Ciências', 'Ciencias', 'CIENCIAS', 'ciências'],
    'História': ['História', 'historia', 'HISTORIA', 'HIST.'],
    'Geografia': ['Geografia', 'GEOGRAFIA', 'geografia', 'GEO.'],
    'Inglês': ['Inglês', 'ingles', 'ING.', 'Língua Inglesa', 'INGLES'],
}
```

O dicionário criado possui a seguinte estrutura: a chave representa o nome da disciplina enquanto o valor é uma lista contendo todas as formas de escrita que representam aquela disciplina.

O objetivo é:
- Fazer uma leitura linha a linha;
- Obter o valor atual na coluna;
- Identificar em qual lista das chaves do dicionário esse valor está;
- Substituir pelo nome da chave, que é a forma padronizada que escolhida para trabalhar neste projeto.

```python
# Utilização do for para obter a chave (Disciplina) e os valores (grafias) no dicionário grafias_disciplinas
for disciplina, grafias in grafias_disciplinas.items():
    # Lê cada tipo de grafia por disciplina, se estiver presente na lista da chave substitui pelo nome da chave.
    # Repete-se esse processo até finalizar a última chave
    for grafia in grafias:
        df_notas['Disciplina'] = df_notas['Disciplina'].replace(
            grafia,
            disciplina
        )
```

```python
df_notas['Disciplina'].unique()
```

```python
df_notas[['Matrícula', 'Nome do Aluno', 'Turma', 'Disciplina']]
```

Os nomes das disciplinas estão formalmente padronizados, facilitando a identificação.

### Conversão das Notas

Diferentemente da conversão de tipos anterior, há algumas coisas que impedem a conversão direta dos dados dessas colunas de notas.

Conforme visto na Análise Exploratória, há alguns valores que são diferentes da estrutura numérica na linguagem Pyhton, além da presença de caracteres textuais. Valores com vírgula (como 8,3) e textos ('sem registro', '-' e '--') são elementos que precisam ser tratados para que seja possível a conversão dos tipos.

#### Números separados por vírgula

Valores separados por vírgula precisam ser trocados por ponto.

A estrutura léxica de valores números com casas decimais em Python é X.Y (exemplo: 1.1, 7.5 e entre outros).

Identificando as strings nas colunas que possuem números com vírgula:

```python
df_notas[
    df_notas['1° Bimestre'].str.contains(',', na=False) |
    df_notas['2° Bimestre'].str.contains(',', na=False) |
    df_notas['3° Bimestre'].str.contains(',', na=False) |
    df_notas['4° Bimestre'].str.contains(',', na=False)
]
```

Há 8 linhas onde existem notas com vírgula.

Aplicando a função de substituição:

```python
colunas_notas = ['1° Bimestre', '2° Bimestre', '3° Bimestre', '4° Bimestre']

#Para cada item dentro desse array, substitui apenas a , pelo . em cada linha da respectiva coluna
for coluna in colunas_notas:
  df_notas[coluna] = df_notas[coluna].str.replace(',', '.', regex=False)
```

```python
df_notas[
    df_notas['1° Bimestre'].str.contains(',', na=False) |
    df_notas['2° Bimestre'].str.contains(',', na=False) |
    df_notas['3° Bimestre'].str.contains(',', na=False) |
    df_notas['4° Bimestre'].str.contains(',', na=False)
]
```

A substituição foi bem sucedida!

#### Textos

Precisamos agora encontrar as linhas que possuem os caracteres textuais.

Para o tratanmento desses valores nesta etapa, a medida adotada foi substituir para NaN. Com isso, já é possível converter para valores numéricos. Posteriormente, os valores nulos serão tratados.

```python
df_notas[
    df_notas['3° Bimestre'].str.contains('-', na=False) |
    df_notas['4° Bimestre'].str.contains('sem registro', na=False) |
    df_notas['4° Bimestre'].str.contains('--', na=False)
]
```

Conforme visto anteriormente na Análise Exploratória, apenas as colunas 3° Bimestre e 4° Bimestre possuem esses caracteres textuais. Apenas 3 linhas possuem essas características.

Aplicando a função de substituição:

```python
# Foi necessária a importação da função nan da biblioteca numpy
from numpy import nan

for coluna in colunas_notas[2:]:
  df_notas[coluna] = df_notas[coluna].replace(['-', 'sem registro', '--'], nan)
```

```python
df_notas[
    df_notas['3° Bimestre'].str.contains('-', na=False) |
    df_notas['4° Bimestre'].str.contains('sem registro', na=False) |
    df_notas['4° Bimestre'].str.contains('--', na=False)
]
```

A substituição foi bem sucedida! Agora é possível realizar as conversões!

#### Convertendo

Com todas as substituições feitas, podemos aplicar a função de alterar os tipos de dados.

```python
# Para cada coluna, converter seu tipo para numérico
for coluna in colunas_notas:
    df_notas[coluna] = pd.to_numeric(df_notas[coluna], errors='coerce')
```

```python
df_notas.info()
```

```python
df_notas[df_notas['Faltas'].isna()]
```

6 alunos estão com as Faltas não registradas. Como tratar esses valores nulos?

A abordagem utilizada neste caso foi: cálculo da média de faltas em outras matérias. O resultado será adicionado no campo de Faltas.

```python
# Visualizando um dos alunos com faltas nulas
df_notas[df_notas['Matrícula'] == 20250133]
```

```python
# Variável
faltas_nulas = df_notas[df_notas['Faltas'].isna()]
```

```python
# Percorre os indices da tabela de faltas nulas
for i in faltas_nulas.index:

  # Obtém o número da matrícula
  matricula = df_notas.loc[i, 'Matrícula']

  # Filtra no df_notas apenas as linhas com a matrícula desejada onde as faltas não são nulas
  faltas_disciplinas = df_notas[
      (df_notas['Matrícula'] == matricula) & (df_notas['Faltas'].notna())
  ]

  # Cálculo da média das faltas
  media_faltas = faltas_disciplinas['Faltas'].mean()

  # Substituição do valor. Arredondando para o inteiro mais próximo
  df_notas.loc[i, 'Faltas'] = round(media_faltas)
```

### Valores nulos em Notas

Para resolver esta situação, é preciso também realizar uma consulta para identificar em quais linhas ocorrem a presença de dados nulos nas notas dos estudantes do 1° ao 4° bimestre.

```python
df_notas[
    df_notas['1° Bimestre'].isna() |
    df_notas['2° Bimestre'].isna() |
    df_notas['3° Bimestre'].isna() |
    df_notas['4° Bimestre'].isna()
]
```

São ao todo 25 linhas onde ocorre o problema de notas nulas. Para resolver esse caso, a estratégia adotada para solucionar é: calcular a média das notas dessa disciplina em outros bimestre e o resultado ocupará a posição nula.

```python
# Variável
notas_nulas = df_notas[
    df_notas['1° Bimestre'].isna() |
    df_notas['2° Bimestre'].isna() |
    df_notas['3° Bimestre'].isna() |
    df_notas['4° Bimestre'].isna()
]
```

```python
# Importação da função isnan do numpy
from numpy import isnan

# Percorre todos os indices da tabela filtrada
for i in notas_nulas.index:
  # Variável para armazenar todas as notas presentes
  lista_notas = []

  # Para cada coluna na lista de colunas referentes as notas
  for coluna in colunas_notas:
    # Valor da nota
    nota = df_notas.loc[i, coluna]

    # Se não for nula, adiciona na lista
    if not isnan(nota):
      lista_notas.append(nota)

  # Calcula a média das notas
  media_notas = round(sum(lista_notas) / len(lista_notas), 1)

  # Para cada coluna na lista de colunas referentes as notas
  for coluna in colunas_notas:
    # Se o valor presente no índice atual e coluna atual for nulo, preenche com a média de notas
    if isnan(df_notas.loc[i, coluna]):
      df_notas.loc[i, coluna] = media_notas
```

### Valores nulos em Situação

Conforme visto no tópico anterior de Padronização, existem 4 linhas com valores nulos em Situação, que estão apresentadas novamente abaixo.

```python
df_notas[df_notas['Situação'].isna()]
```

Os dados referentes as notas e faltas estão completos, portanto podem ser definidas as situações desses alunos realizando os cálculos necessário com base nas Regras de Negócio do projeto.

```python
# Criando uma variável para armazenar as linhas com Situação nula
situacao_nula = df_notas[df_notas['Situação'].isna()]
```

```python
# Aplica-se o for para percorrer as listas de índices das linhas da tabela filtrada
for i in situacao_nula.index:

  # Cálculo da média.
  # Função loc serve para obter o valor com base no índice da linha e nome da coluna
  # Obtém os valores de cada coluna de nota e calcula a média.
  media = (
      df_notas.loc[i, '1° Bimestre'] + df_notas.loc[i, '2° Bimestre'] + df_notas.loc[i, '3° Bimestre'] + df_notas.loc[i, '4° Bimestre']
  ) / 4

  # Cálculo da frequência
  # Obtém o total de aulas e as faltas. Subtrai aulas por faltas e divide para obter o percentual de presença.
  frequencia = ((df_notas.loc[i, 'Aulas'] - df_notas.loc[i, 'Faltas']) / df_notas.loc[i, 'Aulas']) * 100

  # Se média maior ou igual a 6 e frequência maior ou igual a 75%: aprovado.
  # Se média menor que 6: reprovado por nota.
  # Se frequência menor que 75%: reprovado por frequência.
  # Se média e frequência não corresponderem, reprovado por nota e frequência.
  if media >= 6 and frequencia >= 75:
    df_notas.loc[i, 'Situação'] = 'Aprovado'

  elif media < 6 and frequencia >= 75:
    df_notas.loc[i, 'Situação'] = 'Reprovado por Nota'

  elif media >= 6 and frequencia < 75:
    df_notas.loc[i, 'Situação'] = 'Reprovado por Frequência'

  else:
    df_notas.loc[i, 'Situação'] = 'Reprovado por Nota e Frequência'
```
